```python
%%time
#1
# Sum of all numbers divisible by 3 or 5 below 10000

def prob1(limit):
  ttl=0
  for i in range(1,limit):
    if (i%5==0 or i%3==0): ttl+=i
  return ttl

print(prob1(10))
print(prob1(10000))
```

```python
%%time
#2
# Sum of even fibonacci numbers below 4 million

a,b = 1,2
ttl=0
while a<4000000:
  if a%2==0: ttl+=a
  a,b = b, a+b
ttl
```

```python
%%time
#3

def largest_prime(n):
  while n>1:
    for i in range(2,n+1):
      if n%i==0:
        n=n//i
        break
  return i

print(largest_prime(13195))
print(largest_prime(600851475143))
```

```python
%%time
#4

def palindromic(n):
  return str(n)==str(n)[::-1]

max = 0
for i in range(999,100,-1):
  for j in range(999,i+1,-1):
    if palindromic(i*j) and i*j>max:
      max = i*j
print(max)
```

```python
#6
%%time
def sum_of_squares(n):
  ttl=0
  for i in range(1,n+1):
    ttl+=i**2
  return ttl

def square_of_sum(n):
  ttl=0
  for i in range(1,n+1):
    ttl+=i
  return ttl**2

print('10: ',square_of_sum(10)-sum_of_squares(10))
print('100: ',square_of_sum(100)-sum_of_squares(100))
```

```python
%%time
#7

def first_n_primes(n):
  if n==1: return [2]
  primes = [2,3]
  while len(primes)<n:
    i = primes[-1]+2
    while True:
      for j in primes:
        if i%j==0:
          i+=2
          break
      else:
        primes.append(i)
        break
  return primes

print(first_n_primes(6))
print(first_n_primes(10001)[-1])
```

```python
%%time
#10

def primes_below_n(n):
    if n < 2:
        return []
    sieve = [True]*n
    sieve[0:2] = [False, False]  # 0 and 1 are not primes
    for i in range(2, int(n**0.5)+1):
        if sieve[i]:
            sieve[i*i:n:i] = [False]*len(range(i*i, n, i))
    return [i for i, is_prime in enumerate(sieve) if is_prime]

print(sum(primes_below_n(2_000_000)))
```
